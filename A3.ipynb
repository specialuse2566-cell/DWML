```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori , association_rules
```

```python
df = pd.read_csv("Grocery_Dataset.csv")
```

```python
df.head(
)
```

```python
transactions = df.groupby('TransactionID')['Item'].apply(list).tolist()
```

```python
#we now performing the Transaction Encoding and then again to the dataframe for the apriori algo
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)
```

```python
encoded_df = pd.DataFrame(te_array , columns = te.columns_)
encoded_df.head()
```

```python
frequent_itemsets = apriori(
    encoded_df,
    min_support = 0.4,
    use_colnames = True
)
```

```python
frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric = 'confidence',
    min_threshold = 0.6
)
```
